```python
def positions_xyz(track):
    xs=[]
    ys=[]
    zs=[]
    for sample in track:
        xs.append(sample["position_km"][0])
        ys.append(sample["position_km"][1])
        zs.append(sample["position_km"][2])
    return xs, ys, zs
```

```python
xs,ys,zs=positions_xyz(track_a)
print(xs[0],ys[0],zs[0])
```
